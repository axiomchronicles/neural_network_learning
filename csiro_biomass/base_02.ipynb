```python
import torch
import torchvision
import pandas as pd
import matplotlib.pyplot as plt
import pathlib

from sklearn.preprocessing import LabelEncoder
```

```python
root = pathlib.Path(pathlib.Path.home() / "Downloads" / "csiro-biomass")
```

```python
train_data = pd.read_csv(root / "train.csv")
```

```python
train_data.head()
```

```python
sampling_date = pd.to_datetime(train_data["Sampling_Date"])
```

```python
train_data["Sampling_Year"] = sampling_date.dt.year
train_data["Sampling_Month"] = sampling_date.dt.month
train_data["Sampling_Day"] = sampling_date.dt.day
```

```python
train_data
```

```python
state_encoder = LabelEncoder()
species_encoder = LabelEncoder()
```

```python
train_data["State_encoded"] = state_encoder.fit_transform(train_data["State"])
train_data["Species_encoded"] = species_encoder.fit_transform(train_data["Species"])
```

```python
train_data.head()
```

```python
train_data["NDVI_height"] = train_data["Pre_GSHH_NDVI"] ** train_data["Height_Ave_cm"]
train_data["NDVI_sq"] = train_data["Pre_GSHH_NDVI"] ** 2
train_data["Height_sq"] = train_data["Height_Ave_cm"] ** 2
```

```python
train_data.head()
```

```python
train_data["mean_target_per_species"] = train_data.groupby("Species")["target"].transform("mean")
train_data["mean_target_per_state"] = train_data.groupby("State")["target"].transform("mean")
```
