```python
import torch
from torch.utils.data import Dataset
import pandas as pd
from PIL import Image
import pathlib

class BiomassDataset(Dataset):
    def __init__(self, csv_path, root, transform=None, include_tabular=True):
        super().__init__()

        self.annotation = pd.read_csv(csv_path)
        self.root = pathlib.Path(root)
        self.transform = transform
        self.include_tabular = include_tabular

        self.grouped = (
            self.annotation
            .pivot_table(index=[
                "image_path", "Sampling_Date", "State", "Species",
                "Pre_GSHH_NDVI", "Height_Ave_cm"
            ], columns="target_name", values="target")
            .reset_index()
        )

        self.target_cols = [
            "Dry_Clover_g",
            "Dry_Dead_g",
            "Dry_Green_g",
            "Dry_Total_g",
            "GDM_g"
        ]

        if self.include_tabular:
            self._prepare_tabular_features()

    def _prepare_tabular_features(self):
        """Create engineered numeric tabular features."""

        df = self.grouped

        df["Sampling_Date"] = pd.to_datetime(df["Sampling_Date"], errors="coerce")
        df["month"] = df["Sampling_Date"].dt.month.fillna(0)
        df["dayofyear"] = df["Sampling_Date"].dt.dayofyear.fillna(0)

        df["State_enc"] = df["State"].astype("category").cat.codes
        df["Species_enc"] = df["Species"].astype("category").cat.codes

        df["NDVI_height"] = df["Pre_GSHH_NDVI"] * df["Height_Ave_cm"]
        df["NDVI_sq"] = df["Pre_GSHH_NDVI"] ** 2
        df["height_sq"] = df["Height_Ave_cm"] ** 2

        self.tabular_cols = [
            "Pre_GSHH_NDVI", "Height_Ave_cm", "month", "dayofyear",
            "State_enc", "Species_enc", "NDVI_height", "NDVI_sq", "height_sq"
        ]

        self.grouped = df

    def __len__(self):
        return len(self.grouped)

    def __getitem__(self, idx):
        row = self.grouped.iloc[idx]

        img_path = self.root / row["image_path"]
        image = Image.open(img_path).convert("RGB")
        if self.transform:
            image = self.transform(image)

        target = pd.to_numeric(row[self.target_cols], errors="coerce").astype("float32")
        target = torch.tensor(target.values, dtype=torch.float32)

        if self.include_tabular:
            tabular_data = pd.to_numeric(row[self.tabular_cols], errors="coerce").astype("float32")
            tabular_data = torch.tensor(tabular_data.values, dtype=torch.float32)
            return image, tabular_data, target
        else:
            return image, target
```

```python
class BiomassTestDataset(Dataset):
    def __init__(self, csv_path, root, transform = None):
        super(BiomassTestDataset, self).__init__()

        self.annonation = pd.read_csv(csv_path)
        self.root = pathlib.Path(root)
        self.transform = transform

    def __len__(self):
        return len(self.annonation)

    def __getitem__(self, idx):
        rows = self.annonation.iloc[idx]
        
        img_path = self.root / rows["image_path"]
        image = Image.open(img_path).convert("RGB")

        # target = pd.to_numeric(rows["target_name"], errors = "coerce").astype("float32")
        # target = torch.tensor(target).type(torch.float32)

        target = rows["target_name"]
        sample_id = rows["sample_id"]

        # target = torch.tensor(rows["target_name"], dtype=torch.float32)

        if self.transform:
            image = self.transform(image)

        return image, target, sample_id
```

```python
from torchvision import transforms

train_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomVerticalFlip(),
    transforms.RandomRotation(25),
    transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.1),
    transforms.RandomResizedCrop(224, scale=(0.8, 1.0)),
    transforms.ToTensor(),
])

test_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor()
])
```

```python
train_dataset = BiomassDataset(csv_path = root / "train.csv", root = root, transform = train_transform)
test_dataset = BiomassTestDataset(csv_path = root / "test.csv", root = root, transform = test_transform)
```

```python
train_size = int(0.8 * len(train_dataset))
val_size = len(train_dataset) - train_size
train_dataset, val_dataset = torch.utils.data.random_split(train_dataset, [train_size, val_size])

train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False)

image, tabular_data, target = next(iter(train_loader))
image.shape, target.shape, tabular_data.shape
```

```python
import torch
import torch.nn as nn
import torchvision

class CsiroHybridModel(nn.Module):
    def __init__(self, num_tabular_features: int = 9, hidden_units: int = 64, out_features: int = 5):
        super().__init__()

        self.cnn = torchvision.models.resnet18(weights=torchvision.models.ResNet18_Weights.IMAGENET1K_V1)
        in_features = self.cnn.fc.in_features
        self.cnn.fc = nn.Identity() 

        self.tabular = nn.Sequential(
            nn.Linear(num_tabular_features, hidden_units),
            nn.BatchNorm1d(hidden_units),
            nn.ReLU(),
            nn.Dropout(0.25),
            nn.Linear(hidden_units, 32),
            nn.ReLU()
        )

        self.fc = nn.Sequential(
            nn.Linear(in_features + 32, 256),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(256, out_features)
        )

    def forward(self, img, tabular):
        img_feat = self.cnn(img)         # [B, 512]
        tab_feat = self.tabular(tabular) # [B, 32]
        x = torch.cat([img_feat, tab_feat], dim=1)
        return self.fc(x)
```

```python
device = "mps" if torch.mps.is_available() else "cpu"
torch.manual_seed(42)
```

```python
model = CsiroHybridModel().to(device)
# model
```

```python
def r2_score_fn(y_true, y_pred):
    ss_res = torch.sum((y_true - y_pred) ** 2, dim=0)
    ss_tot = torch.sum((y_true - torch.mean(y_true, dim=0)) ** 2, dim=0)
    r2 = 1 - ss_res / ss_tot
    return r2

def weighted_r2(y_true, y_pred):
    weights = torch.tensor([0.1, 0.1, 0.1, 0.5, 0.2], device=y_true.device)
    r2 = r2_score_fn(y_true, y_pred)
    final = torch.sum(weights * r2)
    return r2, final
```

```python
def train_model(model, criterion, optimizer, scheduler, device, dataset, r2score):
    model.train()
    r2score.reset()
    total_loss = 0.0

    for img, tab, target in dataset:
        img, tab, target = img.to(device), tab.to(device), target.to(device)

        optimizer.zero_grad()
        pred = model(img, tab)
        loss = criterion(pred, target)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=2.0)
        optimizer.step()

        total_loss += loss.item()
        r2score.update(pred, target)

    if scheduler:
        scheduler.step()

    avg_loss = total_loss / len(dataset)
    avg_r2 = r2score.compute().item()
    return avg_loss, avg_r2
```

```python
def eval_model(model, criterion, dataset, device):
    model.eval()
    total_loss = 0.0
    y_true, y_pred = [], []

    with torch.inference_mode():
        for img, tab, target in dataset:
            img, tab, target = img.to(device), tab.to(device), target.to(device)
            pred = model(img, tab)
            loss = criterion(pred, target)
            total_loss += loss.item()

            y_true.append(target)
            y_pred.append(pred)

    y_true = torch.cat(y_true)
    y_pred = torch.cat(y_pred)

    r2, weighted = weighted_r2(y_true, y_pred)
    return total_loss / len(dataset), r2.cpu(), weighted.cpu().item()
```

```python
from torchmetrics.regression import R2Score
from tqdm.auto import tqdm

criterion = torch.nn.SmoothL1Loss() 
optimizer = torch.optim.AdamW(model.parameters(), lr=1e-4, weight_decay=1e-4)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=10)

r2score = R2Score().to(device)
```

```python
best_r2 = -float("inf")
patience, trigger = 5, 0  

for epoch in tqdm(range(50), desc="Training"):
    train_loss, train_r2 = train_model(model, criterion, optimizer, scheduler, device, train_loader, r2score)
    val_loss, val_r2s, val_weighted_r2 = eval_model(model, criterion, val_loader, device)

    if epoch % (0.1 * 50) == 0:
        print(f"[Epoch {epoch+1}] Train Loss: {train_loss:.4f}, R²: {train_r2:.4f} | Val Loss: {val_loss:.4f}, Weighted R²: {val_weighted_r2:.4f}")
    
        if val_weighted_r2 > best_r2:
            best_r2 = val_weighted_r2
            torch.save(model.state_dict(), "best_model.pth")
            print("✅ Saved new best model")
```
