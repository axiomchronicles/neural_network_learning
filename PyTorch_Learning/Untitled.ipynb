```python
def count_vowel(abc) :
    count = {}
    # count2 = 0
    vol = ['a', 'i', 'o', 'u' , 'e']
    for idx, n in enumerate(abc):
        if n in vol:
            count[idx] = n
        continue
    return count
print(count_vowel("pokemon is amazing cartoon of the universe"))
print(len(count_vowel("pokemon is amazing cartoon of the universe")))
```
