```python
# Linear Regression

## y = a + bx
# a -> bias (intercept)
# b -> weight (slope)
# x -> independent vairable

weight = 0.7
bias = 0.3

X = torch.arange(0, 10, 0.02).unsqueeze(dim=1)
y = bias + weight * X

splitX = int(0.8 * len(X))
splity = int(0.8 * len(y))

X_train, X_test = X[:splitX], X[splitX:]
y_train, y_test = y[:splity], y[splity:]

# y = a + bx (a: bias(intercepter of the graph), b: weight(sploe of the graph), x: independent(variable))

len(X_train), len(X_test), len(y_train), len(y_test)
```

```python
plt.figure(figsize=(6,4))
plt.plot(X, y, label="y = 0.3 + 0.7x", color="blue")
plt.xlabel("X (independent variable)")
plt.ylabel("y (dependent variable)")
plt.title("Linear Relationship: y = a + b*x")
plt.legend()
plt.grid(True)
plt.show()
```

```python
class LinearRegressionModel(nn.Module):
  def __init__(self):
    super().__init__()
    self.weights = nn.Parameter(torch.randn(1,
                                      requires_grad=True,
                                      dtype=torch.float))
    self.bias = nn.Parameter(torch.randn(1,
                                    requires_grad=True,
                                    dtype=torch.float))

  def forward(self, x: torch.Tensor) -> torch.Tensor:
    return self.bias + self.weights * x
```

```python
torch.manual_seed(42)
X = torch.arange(0, 10, 0.1).unsqueeze(dim=1)
true_weight = 0.7
true_bias = 0.3
y = true_bias + true_weight * X

model = LinearRegressionModel()
with torch.no_grad():
    y_pred = model(X)
```

```python
plt.figure(figsize=(6,4))
plt.scatter(X, y, label="True data", color="blue", s=10)
plt.plot(X, y_pred, label="Model prediction", color="red")
plt.title("Linear Regression Model")
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```

```python
with torch.inference_mode():
  y_pred = model(X_test)

plt.scatter(X, y, label = "Train Data", color= "blue", s=10)
plt.plot(X_test, y_pred, label = "Model Prediction", color = "red")
plt.legend()
plt.show()
```
